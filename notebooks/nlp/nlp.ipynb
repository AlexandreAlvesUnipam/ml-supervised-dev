````python
import pandas as pd
import numpy as np
````

````python
df = pd.read_excel("../data/Customer Call List.xlsx")
df.head()
````

#### Removendo valores duplicados

````python
df[df.duplicated()]
````

````python
df = df.drop_duplicates()
````

#### Removendo coluna desnecesária

````python
df = df.drop(columns = "Not_Useful_Column")
df
````

#### Remove espaços em branco e outros caracteres

````python
df["Last_Name"] = df["Last_Name"].str.strip("123._/") # Argumento opcional
df
````

#### Padronizando valores categóricos

````python
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('Yes','Y')

df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace('No','N')
df
````

#### Lidando com valores faltantes

````python
"""
É necessário avaliar se apenas remover é o suficiente.
Se o volume de nulos for muito alto, essa não é uma boa opção.
"""

df=df.fillna('')
df
````

#### Atualizando índices

````python
df = df.reset_index(drop=True)
df
````
