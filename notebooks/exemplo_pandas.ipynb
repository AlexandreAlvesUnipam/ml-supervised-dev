### Jupyter Notebooks

São ambientes executáveis para scripts python **focados em ciência de dados, machine learning e experimentos**. É comum o seu uso em etapas de teste e validação de novos modelos de machine learning ou pipeline de dados.

Um notebook possui uma estrutura dividida em **células executáveis**. Cada célula pode conter scripts python e ser executada isoladamente. Também é possível escrever markdowns, como no exemplo desta célula.

Por esse motivo é necessário **atenção em relação a ordem de execução das células**.

É bastante flexível e o motivo por ter essa estrutura é justamente para servir como um ambiente de experimentação.

**Menu**: Logo acima é possível visualizar as opções de executar uma célula, criar uma nova célula e reiniciar o kernel. O kernel é ambiente usado pelo notebook para executar localmente, como se fosse um servidor. É necessário escolher qual ambiente python será usado pelo kernel. Assim que você abrir o seu notebook, uma opção estará disponível no canto superior direito com o nome "Select Kernel". 

### Pandas

É a biblioteca mais popular para **extração, análise, processamento, transformação e carregamento de dados**. Possui uma sintaxe intuitiva, o que garante uma inicialização rápida no ambiente. O Pandas possui suporte para praticamente todas as "fontes" de dados, como: arquivos excel (csv, xlsx), bancos de dados, serviços externos de storage e outros.

Por ser uma biblioteca python, é muito utilizada para a criação de **pipelines de dados e scripts de I.A e machine learning**. Possui uma estrutura de dados própria, chamada de **DataFrame** que será vista a seguir.

```python
import pandas as pd
```

### Extração

```python
'''
Exemplo de carregamento de um conjunto de dados em formato .csv
'''

iris = pd.read_csv("../data/flores_iris.csv")
```

```python
'''
Visualizando os primeios dados da planilha
'''
iris
```

**Obervações**: O separador do seu arquivo pode ser vírgula (,) ou ponto e vírgula (;). Por padrão o `read_csv()` utiliza a vírgula. Mas existe também a opção `read_xlsx()` que possui como padrão o ponto e vígular. Caso o arquivo seja um .csv mas mesmo assim utilize o ponto é vírgula, é possível usar um parâmetro que informa o separador.

`read_csv('flores_iris.csv', sep=';')`

**Documentação do pandas**: https://pandas.pydata.org/docs/reference/index.html

### DataFrame

A extração feita anteriormente com o arquivo csv gerou um **objeto chamado DataFrame**. É a estrutura de dados usada pelo pandas e que representa uma tabela. Todas as operações são feitas nessas tabelas. O DataFrame é composto por outros objetos chamados de **Series**, que representam as colunas da tabela.

É possível criar um DataFrame usando dicionários do python:

```python
'''
O dicionário representa o DataFrame.
Cada lista dentro do dicionário representa uma Serie.
'''
df = pd.DataFrame(
    {
        "Name": [
            "Fulano da Silva",
            "Ciclano José"
        ],
        "Idade": [35, 44],
        "Profissão": ["Programador", "Analista de Dados"]
    }
)
```

```python
df
```

```python
'''Visualizando os tipos'''
print(type(df))
print(type(df['Name']))#
```

Na sintaxe do Pandas usamos os colchetes "[]" para retornar uma Series. Também é usado para filtragem e outras operações como vamos ver mais a frente.

A partir de uma Series podemos chamar métodos para extrair informações. [Métodos disponíveis](https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series)

```python
# Retorna o maior valor da coluna
#df["Idade"].max()
# Número de valores não nulos
#df["Idade"].count()
# Estatísticas gerais
df["Idade"].describe()
```

```python
df.dtypes
```

```python
# Selecionando subconjunto de colunas
sample = iris[['SepalLengthCm', 'SepalWidthCm']]
sample
```

```python
# Filtrando DataFrame usando uma condição
sample_filter = iris[iris['SepalLengthCm']>5]
sample_filter
```

```python
#Obetendo apenas a coluna mas com o mesmo filtro
col = iris['SepalLengthCm']>5
col
```

```python
# Obtendo média dos valores de uma coluna
iris['SepalLengthCm'].mean()
```

```python
# iloc é um método onde buscamos uma linha pelo índice
sample_iloc = iris.iloc[3]
sample_iloc
```

```python
type(sample_iloc)
```

```python
# Já o loc nos permite passar linha e coluna, fazendo filtros mais complexos
sample_loc = iris.loc[(iris['SepalLengthCm']>5) & (iris['SepalWidthCm']>3)]
sample_loc
```

```python
'''
DataSet filtrado com apenas comprimento e largura da sépala.
Usando método describe() para obter as estatísticas dessas variáveis
'''
iris[['SepalLengthCm', 'SepalWidthCm']].describe()
```

```python
'''
value_counts(): Volume de dados para cada valor/classe
value_counts(normalize=True): Em formato de porcentagem
'''
iris['Species'].value_counts(normalize=True)
```

```python
'''
Renomeando as colunas
Retorna um novo DataFrame se o parâmetro 'inplace=False'
'''

exportar_setosa = iris.rename(
    columns={
        "SepalLengthCm": "ComprimentoSepala",
        "SepalWidthCm": "LarguraSepala",
        "PetalLengthCm": "ComprimentoPetala",
        "PetalWidthCm": "LarguraPetala",
        "Species": "Classe"
        })
exportar_setosa
```

```python
'''Apenas Iris Setosa'''
exportar_setosa['Classe']=="Iris-setosa"
```

```python
exportar_setosa = exportar_setosa[exportar_setosa['Classe']=="Iris-setosa"]
```

```python
'''Exporta para csv'''
exportar_setosa.to_csv("iris_setosa.csv")
```
