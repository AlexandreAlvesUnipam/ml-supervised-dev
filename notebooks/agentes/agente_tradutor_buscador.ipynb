```python
import os
from langchain_mistralai import ChatMistralAI
from dotenv import load_dotenv

from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.prompts import PromptTemplate

load_dotenv()
```

```python
api_key = os.getenv("MISTRAL_API_KEY")
if not api_key:
    print("API KEY não fornecida")
```

Definindo modelo LLM

```python
'''
Cria objeto para consumir modelos do Mistral. Outros provedores são bem parecidos.
Link da documentação: https://python.langchain.com/api_reference/mistralai/chat_models/langchain_mistralai.chat_models.ChatMistralAI.html#langchain_mistralai.chat_models.ChatMistralAI.get_num_tokens_from_messages
'''

llm = ChatMistralAI(
    model="open-mistral-7b",
    temperature=0,
    max_retries=1,
    verbose=True # Ajuda a ver tokens e modelo
)
```

Definindo a tool

```python
search = DuckDuckGoSearchRun()
```

Definindo prompt

```python
'''
O PromptTemplate permite a formatação dinâmica de nossos prompts. Muito comum
quando temos muitos prompts para vários cenários e esses são organizados em arquivos
separados, como por exemplo no formato .jinja2
'''

prompt = PromptTemplate(
    input_variables=["text"],
    template="Traduza o seguinte texto para português brasileiro, mantendo o significado original:\n\n{text}"
)
```

Usando chain com o operador pipe '|'

```python
def create_search_and_translate_chain(query):
    search = DuckDuckGoSearchRun()
    search_result = search.invoke(query)
    
    translation_chain = (
        {"text": lambda x: x}
        | prompt 
        | llm
    )
    
    return translation_chain.invoke(search_result)

if __name__ == "__main__":
    query = "What is the capital of France?"
    result = create_search_and_translate_chain(query)
    print("Resultado traduzido:")
    print(result)
```
